```python
import pandas as pd

# ระบุ path ของไฟล์ CSV ต้นทาง
input_file_path = 'dataset.csv'  # แก้ไขเป็น path ของคุณ

# อ่านไฟล์ CSV ทั้งหมด
df = pd.read_csv(input_file_path)

# ตรวจสอบจำนวน record ทั้งหมด
total_records = len(df)
print(f"จำนวน record ทั้งหมด: {total_records}")

# กำหนดจำนวน record สำหรับแต่ละชุด
train_size = 600000
test_size = 300000

# ตรวจสอบว่าจำนวน record ตรงตามที่ต้องการหรือไม่
if total_records < (train_size + test_size):
    print("Error: จำนวน record ในไฟล์ต้นทางไม่เพียงพอ")
else:
    # แยก DataFrame ออกเป็น 2 ชุด
    df_train = df.iloc[:train_size].copy()  # 600,000 record แรก
    df_test = df.iloc[train_size:train_size + test_size].copy()  # 300,000 record ถัดไป

    # บันทึกเป็นไฟล์ CSV ใหม่
    df_train.to_csv('training_set.csv', index=False)
    df_test.to_csv('test_set.csv', index=False)

    print("แยกไฟล์เสร็จสิ้น")
    print(f"บันทึก training_set.csv จำนวน {len(df_train)} records")
    print(f"บันทึก test_set.csv จำนวน {len(df_test)} records")
```
