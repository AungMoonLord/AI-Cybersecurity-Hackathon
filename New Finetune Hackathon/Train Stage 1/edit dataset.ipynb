```python
def fake_wrap(text):
    wrap = f"timestamp=2025-01-01 00:00:00 user=webapp ip=0.0.0.0 duration=0.01 sql={text}"
    return wrap
```

```python
fake_wrap("SELECT id FROM orders WHERE status='pending' OR 1=1")
```

```python
import pandas as pd

# สร้างตัวอย่าง DataFrame (แทนที่คุณจะโหลดจากไฟล์หรือ Google Sheet)
df = pd.read_csv('SQL in fake log wrapper.csv')

print("=== ก่อนแก้ไข ===")
print(df)

# แก้ไขทีละ row ในคอลัมน์ 'query log'
for i in range(len(df)):
    original_text = df.loc[i, 'query log']
    # ตัวอย่าง: เปลี่ยนให้เป็นตัวพิมพ์ใหญ่ หรือเพิ่มคำนำหน้า
    modified_text = fake_wrap(original_text)
    df.loc[i, 'query log'] = modified_text

print("\n=== หลังแก้ไข ===")
print(df)

df.to_csv('SQL in fake log wrapper - modified.csv', index=False)
```
